What is the most common color out of all mushrooms

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv("mushrooms.csv")

counts = (
    df[["cap-color", "odor"]]
    .value_counts()
    .reset_index(name="count")
)

sns.barplot(
    data=counts,
    x="cap-color",
    y="count",
    hue="odor"
)

plt.title("Mushroom Cap Color and Odor Counts")
plt.xlabel("Cap Color")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
